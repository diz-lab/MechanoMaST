```python
genes=['COL1A1','COL1A2','COL3A1','COL12A1','COL5A1']

for gene in genes:
    # Prepare color map
    samples = adata.obs['sample'].unique()
    colors = plt.cm.tab20(range(len(samples)))
    color_map = dict(zip(samples, colors))
    hex_colors = [mcolors.to_hex(color_map[sample]) for sample in samples]
    palette = dict(zip(samples, hex_colors))

    # Get gene ID for TGFBI
    gene_id = adata.var_names[adata.var['SYMBOL'] == gene][0]

    # Extract expression data
    expr_values = adata[:, gene_id].X
    if hasattr(expr_values, "todense"):
        expr_values = expr_values.todense()
    values_flat = expr_values.A1 if hasattr(expr_values, "A1") else expr_values.flatten()

    # Build DataFrame
    df = pd.DataFrame({
        gene: values_flat,
        'sample': adata.obs['sample'].values
    })

    # Figure dimensions
    fig_width_mm = 43.5
    fig_width_inch = fig_width_mm / 25.4
    fig_height_inch = fig_width_inch * 0.8

    # Plot
    fig, ax = plt.subplots(figsize=(fig_width_inch, fig_height_inch), dpi=600)
    sns.violinplot(
        data=df,
        x='sample',
        y=gene,
        order=samples,
        palette=palette,
        scale='width',
        inner='quart',
        cut=0,
        linewidth=0.3,
        width=0.8,
        ax=ax
    )

    # Style the plot
    ax.spines['right'].set_visible(False)
    ax.spines['top'].set_visible(False)
    ax.spines['left'].set_linewidth(0.3)
    ax.spines['bottom'].set_linewidth(0.3)

    ax.set_xticklabels(
        ax.get_xticklabels(),
        rotation=45,
        ha='right',
        rotation_mode='anchor',
        fontsize=5
    )

    ax.tick_params(width=0.3, labelsize=5)
    plt.xticks(rotation=45, ha='right')
    plt.xlabel('')
    plt.ylabel(f'{gene} expression', fontsize=5)

    plt.tight_layout()
    plt.savefig(f'/g/korbel/olisov/Visium_Linda/paper_figures/11samples/{gene}_violin_by_sample.svg', dpi=600)
    plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import linregress
import numpy as np

# Load data
df = pd.read_csv('/g/korbel/olisov/Visium_Linda/stiffness/normalized/merged_11samples_50um_final.tsv', sep='\t', index_col=0)

# Apply log10 transform to 'mean' after filtering
df = df[df['mean'] >= 10]
df['mean'] = np.log10(df['mean'])

gene = 'TGFBI'
df = df[(df[gene] > 0)]

# Set fixed figure dimensions
fig_width_mm = 43.5
fig_width_inch = fig_width_mm / 25.4
fig_height_inch = fig_width_inch * 0.8

# Create figure and axis
fig, ax = plt.subplots(figsize=(fig_width_inch, fig_width_inch), dpi=600)

# Scatter plot
sns.scatterplot(data=df, x=gene, y='mean', color='grey', s=3, ax=ax)

# Correlation line (in linear space since mean is already log-transformed)
slope, intercept, r_value, p_value, std_err = linregress(df[gene], df['mean'])
x_sorted = np.linspace(df[gene].min(), df[gene].max(), 500)
y_sorted = intercept + slope * x_sorted
ax.plot(x_sorted, y_sorted, color='red', linewidth=0.5)

# Annotate r and p value
ax.text(0.5, 0.95, f'$r$ = {r_value:.2f}', transform=ax.transAxes,
        ha='center', va='bottom', fontsize=3)
ax.text(0.5, 0.90, f'$p$ = {p_value:.2e}', transform=ax.transAxes,
        ha='center', va='bottom', fontsize=3)

# Labels and formatting
ax.set_xlabel(f'{gene} expression', fontsize=3, labelpad=1)
ax.set_ylabel('log₁₀Stiffness', fontsize=3, labelpad=1)
ax.tick_params(axis='both', which='major', pad=0.5, width=0.3, labelsize=3)

# Axis limits and ticks
x_min, x_max = df[gene].min(), df[gene].max()
y_min, y_max = 1.4, df['mean'].max()

ax.set_xticks(np.arange(np.floor(x_min), np.ceil(x_max), 1))
ax.set_yticks(np.arange(np.floor(y_min), np.ceil(y_max), 1))
ax.set_ylim(bottom=1.4)
ax.set_xlim(left=0.7)

sns.despine(top=True, right=True)
plt.tight_layout()
plt.savefig(f'/g/korbel/olisov/Visium_Linda/paper_figures/11samples/{gene}_correlation.svg', dpi=600, bbox_inches='tight')
plt.show()
```
